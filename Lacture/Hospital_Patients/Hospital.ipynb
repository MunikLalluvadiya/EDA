UNIVARATE

Top 4 Variables to Focus On: Treatment_Cost_USD, Length_of_Stay_Days, Department, and Admission_Type.

```python
# fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(df["Treatment_Cost_USD"], kde=True, bins=30)

plt.title("Distribution of Treatment Cost (USD)")
plt.xlabel("Cost Of Patient")
```

```python
sns.boxplot(data=df, x=df["Length_of_Stay_Days"])

plt.title("Length of Stay (Days) Boxplot")
plt.xlabel("Days")
```

```python
sns.countplot(data=df, y="Department", order=df["Department"].value_counts().index)

plt.title("Patient Volume by Department")
plt.xlabel("Number Of Patient")
```

```python
sns.countplot(data=df, y="Admission_Type", order=df["Admission_Type"].value_counts().index)

plt.title("Patient Admissions by Urgency Type")
plt.xlabel("Number Of Patient")
```

Bivariate Analysis

```python

plt.figure(figsize=(10,6))
numeric_cols = ["Age","BMI","Blood_Pressure_Systolic","Cholesterol_Level","Heart_Rate","Length_of_Stay_Days","Treatment_Cost_USD"]

Corr_Num_Clm = df[numeric_cols].corr()

sns.heatmap(Corr_Num_Clm,annot=True)      #, vmin=-1, vmax=1

plt.title("Correlation Heatmap of Clinical & Financial Metrics")
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="Length_of_Stay_Days", y="Treatment_Cost_USD", hue="Admission_Type")

plt.title("Treatment Cost vs. Length of Stay by Admission Type")
plt.xlabel("Length of Stay (Days)")
plt.ylabel("Treatment Cost ($)")
```

```python
fig, axes = plt.subplots(1, 2 , figsize=(15,5))

sns.boxplot(data=df, x="Department", y="Treatment_Cost_USD", ax=axes[0])

axes[0].tick_params(axis="x", rotation=45)
axes[0].set_title("Treatment Cost Variation Across Departments")



sns.boxplot(data=df, x="Admission_Type", y="Treatment_Cost_USD", ax=axes[1])

axes[1].set_title("Treatment Cost Across Admission Types")
```
